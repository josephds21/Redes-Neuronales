```python
# PRONÓSTICO DE TRÁFICO VEHICULAR

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense
from tensorflow.keras.optimizers import Adam

print("TensorFlow:", tf.__version__)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)

URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00492/Metro_Interstate_Traffic_Volume.csv.gz"

df = pd.read_csv(URL)

print("\nDataset cargado correctamente.")
print("Dimensiones del dataset:", df.shape)

print("\nPrimeros registros:")
print(df.head())

print("\nInformación del dataset:")
df.info()

print("\nColumnas disponibles:")
print(df.columns.tolist())

print("\nValores faltantes:")
print(df.isnull().sum())

datos = df[["date_time", "traffic_volume"]].copy()

datos.columns = ["Fecha", "Trafico"]

datos["Fecha"] = pd.to_datetime(datos["Fecha"])

datos = datos.sort_values("Fecha")

datos = datos.dropna()

datos = datos.reset_index(drop=True)

print("\nDatos después del preprocesamiento:")
print(datos.head())

print("\nInformación:")
datos.info()

print("\nFecha inicial:", datos["Fecha"].min())
print("Fecha final:", datos["Fecha"].max())

print("Cantidad de registros:", len(datos))

print("\nEstadísticas del tráfico:")
print(datos["Trafico"].describe())

plt.figure(figsize=(14, 5))

plt.plot(
    datos["Fecha"],
    datos["Trafico"]
)

plt.title("Volumen de tráfico vehicular")
plt.xlabel("Fecha")
plt.ylabel("Vehículos por hora")

plt.grid()
plt.tight_layout()

plt.show()

inicio = datos["Fecha"].min()

fin = inicio + pd.Timedelta(days=7)

semana = datos[
    (datos["Fecha"] >= inicio) &
    (datos["Fecha"] < fin)
]

plt.figure(figsize=(14, 5))

plt.plot(
    semana["Fecha"],
    semana["Trafico"],
    marker="o"
)

plt.title("Volumen de tráfico durante una semana")
plt.xlabel("Fecha y hora")
plt.ylabel("Vehículos por hora")

plt.xticks(rotation=45)

plt.grid()
plt.tight_layout()

plt.show()

scaler = MinMaxScaler(
    feature_range=(0, 1)
)

trafico_escalado = scaler.fit_transform(
    datos[["Trafico"]]
)

print("\nPrimeros valores normalizados:")
print(trafico_escalado[:10])

porcentaje_entrenamiento = 0.80

indice_entrenamiento = int(
    len(trafico_escalado) *
    porcentaje_entrenamiento
)

train_data = trafico_escalado[
    :indice_entrenamiento
]

test_data = trafico_escalado[
    indice_entrenamiento:
]

print("\nDatos de entrenamiento:", len(train_data))
print("Datos de prueba:", len(test_data))

# Utilizaremos las últimas 24 horas
# para predecir la siguiente hora.

ventana = 24

def crear_secuencias(data, ventana):

    X = []
    y = []

    for i in range(
        ventana,
        len(data)):

        X.append(
            data[i - ventana:i, 0] )

        y.append(data[i, 0])

    return ( np.array(X), np.array(y) )

X_train, y_train = crear_secuencias(
    train_data,
    ventana
)

print("\nDimensiones de entrenamiento:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

datos_prueba = np.concatenate(
    ( train_data[-ventana:], test_data ))

X_test, y_test = crear_secuencias(
    datos_prueba,
    ventana)

print("\nDimensiones de prueba:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

X_train = X_train.reshape(
    X_train.shape[0],
    X_train.shape[1],
    1)

X_test = X_test.reshape(
    X_test.shape[0],
    X_test.shape[1],
    1)

print("\nDimensiones finales:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

modelo = Sequential([

    Input(
        shape=(ventana, 1)
    ),

    LSTM( 50, return_sequences=True ), LSTM( 30 ), Dense( 1)])

modelo.compile( optimizer=Adam( learning_rate=0.001 ), loss="mean_squared_error")

print("\nArquitectura del modelo:")

modelo.summary()

historial = modelo.fit( X_train,  y_train, epochs=50, batch_size=32, validation_split=0.1, verbose=1)

plt.figure(figsize=(8, 5))

plt.plot(
    historial.history["loss"],
    label="Entrenamiento"
)

plt.plot(
    historial.history["val_loss"],
    label="Validación"
)

plt.xlabel("Época")
plt.ylabel("MSE")

plt.title(
    "Curva de aprendizaje de la LSTM"
)

plt.legend()

plt.grid()

plt.tight_layout()

plt.show()

predicciones = modelo.predict(
    X_test,
    verbose=0
)

print(
    "\nCantidad de predicciones:",
    len(predicciones)
)

predicciones_reales = scaler.inverse_transform(
    predicciones
)

y_test_reales = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)


print("\nPrimeras predicciones:")

print(
    predicciones_reales[:10]
)

print("\nPrimeros valores reales:")

print(
    y_test_reales[:10]
)

mae = mean_absolute_error(
    y_test_reales,
    predicciones_reales
)

mse = mean_squared_error(
    y_test_reales,
    predicciones_reales
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test_reales,
    predicciones_reales
)

print("\n===================================")
print("       MÉTRICAS DEL MODELO")
print("===================================")

print(
    f"MAE:  {mae:.2f}"
)

print(
    f"MSE:  {mse:.2f}"
)

print(
    f"RMSE: {rmse:.2f}"
)

print(
    f"R²:   {r2:.4f}"
)

plt.figure(figsize=(14, 5))

plt.plot( y_test_reales, label="Valor real")

plt.plot( predicciones_reales, label="Valor predicho")

plt.title( "Tráfico real vs. tráfico predicho")

plt.xlabel("Hora")

plt.ylabel( "Vehículos por hora")

plt.legend()

plt.grid()

plt.tight_layout()

plt.show()

horas_grafica = 168

plt.figure(figsize=(14, 5))

plt.plot(
    y_test_reales[ -horas_grafica: ], label="Real")

plt.plot( predicciones_reales[ -horas_grafica: ], label="Predicho")

plt.title( "Tráfico real vs. predicho - Última semana")

plt.xlabel("Hora")

plt.ylabel( "Vehículos por hora")

plt.legend()

plt.grid()

plt.tight_layout()

plt.show()


resultados = pd.DataFrame({

    "Trafico_Real":
        y_test_reales.flatten(),

    "Trafico_Predicho":
        predicciones_reales.flatten()

})

resultados["Error"] = ( resultados["Trafico_Real"] - resultados["Trafico_Predicho"])


print("\nPrimeros resultados:")

print( resultados.head(20))

ultima_ventana = trafico_escalado[ -ventana:].copy()

pronostico = []

ventana_actual = ultima_ventana.copy()

for i in range(24):

    entrada = ventana_actual.reshape( 1, ventana, 1 )

    prediccion = modelo.predict( entrada, verbose=0 )[0, 0]

    pronostico.append( prediccion)

    ventana_actual = np.append( ventana_actual[1:], [[prediccion]], axis=0 )

pronostico_real = scaler.inverse_transform(
    np.array( pronostico ).reshape(-1, 1))

pronostico_real = ( pronostico_real.flatten())

print("\n===================================")
print("   PRONÓSTICO PRÓXIMAS 24 HORAS")
print("===================================")

for i, valor in enumerate( pronostico_real, start=1):

    print( f"Hora +{i}: "
        f"{valor:.0f} vehículos")


ultima_fecha = datos[
    "Fecha"
].iloc[-1]

fechas_futuras = pd.date_range(

    start=
        ultima_fecha
        +
        pd.Timedelta(hours=1),
    periods=24,
    freq="h"
)

pronostico_df = pd.DataFrame({

    "Fecha":
        fechas_futuras,
    "Trafico_Pronosticado":
        pronostico_real})

print( "\nPronóstico:")

print( pronostico_df)

plt.figure(figsize=(14, 5))

plt.plot(
    datos["Fecha"].iloc[-168:],
    datos["Trafico"].iloc[-168:],
    label="Histórico")

plt.plot(
    fechas_futuras,
    pronostico_real,
    marker="o",
    label="Pronóstico")

plt.title( "Pronóstico del tráfico vehicular para las próximas 24 horas")

plt.xlabel( "Fecha y hora")

plt.ylabel( "Vehículos por hora")

plt.xticks( rotation=45)

plt.legend()

plt.grid()

plt.tight_layout()

plt.show()

pronostico_df.to_csv( "pronostico_trafico_24_horas.csv", index=False)

print( "\nArchivo 'pronostico_trafico_24_horas.csv' " "guardado correctamente.")

resultados.to_csv("resultados_trafico_lstm.csv", index=False)
```
