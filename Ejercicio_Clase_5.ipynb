```python
# EJERCICIO - CLASIFICACIÓN CON RED NEURONAL
# DATASET WINE

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input


wine = load_wine()

df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

df["target"] = wine.target

print("Primeros registros del dataset:")
print(df.head())


print("\n========================================")
print("INFORMACIÓN DEL DATASET")
print("========================================")

print("Tamaño del dataset:", df.shape)

print("\nInformación:")
df.info()

print("\nEstadísticas descriptivas:")
print(df.describe())


print("\n========================================")
print("LIMPIEZA DE DATOS")
print("========================================")

# Revisar valores nulos
print("\nValores nulos:")
print(df.isnull().sum())

# Revisar duplicados
print("\nDuplicados antes de eliminar:")
print(df.duplicated().sum())

# Eliminar duplicados
df = df.drop_duplicates()

print("\nDuplicados después de eliminar:")
print(df.duplicated().sum())


print("\n========================================")
print("DISTRIBUCIÓN DE CLASES")
print("========================================")

print("Clases existentes:")
print(df["target"].unique())

print("\nCantidad de registros por clase:")
print(df["target"].value_counts().sort_index())

print("\nNombre de las clases:")

for i, nombre in enumerate(wine.target_names):
    print(i, "=", nombre)


plt.figure(figsize=(6, 4))

df["target"].value_counts().sort_index().plot(kind="bar")

plt.xlabel("Clase")
plt.ylabel("Cantidad de registros")
plt.title("Distribución de clases del dataset Wine")
plt.xticks(
    [0, 1, 2],
    wine.target_names,
    rotation=0
)

plt.show()


X = df.drop("target", axis=1)
y = df["target"]

print("\n========================================")
print("CARACTERÍSTICAS Y ETIQUETAS")
print("========================================")

print("Número de características:", X.shape[1])
print("Número de registros:", X.shape[0])

print("\nPrimeros valores de X:")
print(X.head())

print("\nPrimeros valores de y:")
print(y.head())


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n========================================")
print("DIVISIÓN DE DATOS")
print("========================================")

print("Datos de entrenamiento:", X_train.shape)
print("Datos de prueba:", X_test.shape)


scaler = StandardScaler()

# El scaler aprende SOLO de los datos de entrenamiento
X_train = scaler.fit_transform(X_train)

# Al conjunto de prueba solamente se le aplica la transformación
X_test = scaler.transform(X_test)

print("\n========================================")
print("ESTANDARIZACIÓN")
print("========================================")

print("Primer registro estandarizado:")
print(X_train[0])


modelo = Sequential([
    Input(shape=(13,)),
    Dense(8, activation="relu"),
    Dense(8, activation="relu"),
    Dense(3, activation="softmax")
])

print("\n========================================")
print("ARQUITECTURA DEL MODELO")
print("========================================")

modelo.summary()


modelo.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


print("\n========================================")
print("ENTRENAMIENTO")
print("========================================")

historial = modelo.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.20,
    verbose=1
)


plt.figure(figsize=(8, 5))

plt.plot(
    historial.history["loss"],
    label="Entrenamiento"
)

plt.plot(
    historial.history["val_loss"],
    label="Validación"
)

plt.xlabel("Época")
plt.ylabel("Loss")
plt.title("Curva de aprendizaje - Pérdida")

plt.legend()
plt.show()


plt.figure(figsize=(8, 5))

plt.plot(
    historial.history["accuracy"],
    label="Entrenamiento"
)

plt.plot(
    historial.history["val_accuracy"],
    label="Validación"
)

plt.xlabel("Época")
plt.ylabel("Accuracy")
plt.title("Curva de aprendizaje - Accuracy")

plt.legend()
plt.show()


print("\n========================================")
print("CLASIFICACIÓN")
print("========================================")

# Obtener probabilidades
predicciones = modelo.predict(X_test)

# Obtener la clase con mayor probabilidad
y_pred = np.argmax(predicciones, axis=1)

print("Predicciones:")
print(y_pred)

print("\nValores reales:")
print(y_test.to_numpy())


accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n========================================")
print("RESULTADO DE CLASIFICACIÓN")
print("========================================")

print("Accuracy:", accuracy)

print("Accuracy (%):", accuracy * 100)


matriz = confusion_matrix(
    y_test,
    y_pred
)

print("\n========================================")
print("MATRIZ DE CONFUSIÓN")
print("========================================")

print(matriz)


plt.figure(figsize=(7, 6))

plt.imshow(matriz)

plt.title("Matriz de Confusión")
plt.xlabel("Predicción")
plt.ylabel("Valor real")

plt.xticks(
    [0, 1, 2],
    wine.target_names
)

plt.yticks(
    [0, 1, 2],
    wine.target_names
)

# Mostrar números dentro de la matriz
for i in range(3):
    for j in range(3):
        plt.text(
            j,
            i,
            matriz[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()

plt.show()

print("\n========================================")
print("REPORTE DE CLASIFICACIÓN")
print("========================================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=wine.target_names
    )
)


print("\n========================================")
print("RESUMEN FINAL")
print("========================================")

print("Número de registros:", len(df))
print("Número de características:", X.shape[1])
print("Número de clases:", len(wine.target_names))

print("\nClases:")

for i, nombre in enumerate(wine.target_names):
    print(i, "=", nombre)

print("\nAccuracy final:", round(accuracy * 100, 2), "%")
```
