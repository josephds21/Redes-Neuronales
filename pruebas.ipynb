```python
import numpy as np
import matplotlib.pyplot as plt


# ==========================================
# 1. FUNCIÓN DE ACTIVACIÓN
# ==========================================

def funcion_activacion(valor):
    if valor >= 0:
        return 1
    else:
        return 0


# ==========================================
# 2. FUNCIÓN DE PREDICCIÓN
# ==========================================

def predecir(entradas, pesos, sesgo):
    suma_ponderada = np.dot(entradas, pesos) + sesgo
    salida = funcion_activacion(suma_ponderada)

    return salida


# ==========================================
# 3. DATOS DE ENTRENAMIENTO
# COMPuerta AND
# ==========================================

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])


# ==========================================
# 4. PARÁMETROS INICIALES
# ==========================================

tasa_aprendizaje = 0.1
epocas = 10


# ==========================================
# 5. FUNCIÓN DE ENTRENAMIENTO
# ==========================================

def entrenar_perceptron(X, y, tasa_aprendizaje, epocas):

    pesos = np.zeros(X.shape[1])
    sesgo = 0.0

    errores_por_epoca = []

    for epoca in range(epocas):

        error_total = 0

        for i in range(len(X)):

            prediccion = predecir(X[i], pesos, sesgo)

            error = y[i] - prediccion

            pesos = pesos + tasa_aprendizaje * error * X[i]

            sesgo = sesgo + tasa_aprendizaje * error

            error_total = error_total + abs(error)

        errores_por_epoca.append(error_total)

        print("Época:", epoca + 1)
        print("Error total:", error_total)
        print("Pesos:", pesos)
        print("Sesgo:", sesgo)
        print("-------------------------")


    return pesos, sesgo, errores_por_epoca


# ==========================================
# 6. ENTRENAR EL PERCEPTRÓN
# ==========================================

pesos_finales, sesgo_final, errores = entrenar_perceptron(
    X,
    y,
    tasa_aprendizaje,
    epocas
)


# ==========================================
# 7. MOSTRAR PARÁMETROS FINALES
# ==========================================

print("\n==============================")
print("RESULTADOS FINALES")
print("==============================")

print("Pesos finales:", pesos_finales)
print("Sesgo final:", sesgo_final)


# ==========================================
# 8. PROBAR EL MODELO
# ==========================================

print("\n==============================")
print("PRUEBAS DEL MODELO ENTRENADO")
print("==============================")

for entrada in X:

    salida = predecir(
        entrada,
        pesos_finales,
        sesgo_final
    )

    print(
        "Entrada:",
        entrada,
        "Predicción:",
        salida
    )


# ==========================================
# 9. PRUEBAS INDIVIDUALES
# ==========================================

entrada_1 = np.array([0, 0])
entrada_2 = np.array([0, 1])
entrada_3 = np.array([1, 0])
entrada_4 = np.array([1, 1])

print("\n==============================")
print("PREDICCIONES INDIVIDUALES")
print("==============================")

print(
    "Entrada [0, 0]:",
    predecir(entrada_1, pesos_finales, sesgo_final)
)

print(
    "Entrada [0, 1]:",
    predecir(entrada_2, pesos_finales, sesgo_final)
)

print(
    "Entrada [1, 0]:",
    predecir(entrada_3, pesos_finales, sesgo_final)
)

print(
    "Entrada [1, 1]:",
    predecir(entrada_4, pesos_finales, sesgo_final)
)


# ==========================================
# 10. GRÁFICA DEL ERROR
# ==========================================

plt.plot(
    range(1, epocas + 1),
    errores,
    marker='o'
)

plt.title("Error durante el entrenamiento")
plt.xlabel("Época")
plt.ylabel("Error total")
plt.grid(True)
plt.show()
```

```python
# PRÁCTICA 4: PRONÓSTICO CON REDES NEURONALES
# RNN SIMPLE IMPLEMENTADA DESDE CERO

import numpy as np
import matplotlib.pyplot as plt

tiempo = np.linspace(0, 20, 200)
serie = np.sin(tiempo)

plt.figure(figsize=(10, 4))

plt.plot(tiempo, serie)

plt.title("Serie temporal original")
plt.xlabel("Tiempo")
plt.ylabel("Valor")
plt.grid(True)

plt.show()

def crear_secuencias(serie, pasos):

    X = []
    y = []

    for i in range(len(serie) - pasos):

        X.append(serie[i:i + pasos])
        y.append(serie[i + pasos])

    return np.array(X), np.array(y)

pasos = 5

X, y = crear_secuencias(serie, pasos)

# Darle a X la forma necesaria para la RNN
X = X.reshape(X.shape[0], X.shape[1], 1)

# Darle a y forma de columna
y = y.reshape(-1, 1)

print("Forma de X:", X.shape)
print("Forma de y:", y.shape)

def tanh(x):
    return np.tanh(x)


def derivada_tanh(x):
    return 1 - x ** 2

np.random.seed(42)

neuronas_entrada = 1
neuronas_ocultas = 8
neuronas_salida = 1


# Pesos entre entrada y capa oculta
Wxh = np.random.randn(
    neuronas_entrada,
    neuronas_ocultas
) * 0.1


# Pesos recurrentes
Whh = np.random.randn(
    neuronas_ocultas,
    neuronas_ocultas
) * 0.1


# Pesos entre capa oculta y salida
Why = np.random.randn(
    neuronas_ocultas,
    neuronas_salida
) * 0.1


# Sesgo de la capa oculta
bh = np.zeros(
    (1, neuronas_ocultas)
)


# Sesgo de salida
by = np.zeros(
    (1, neuronas_salida)
)

print("\nDimensiones de los parámetros:")
print("Wxh:", Wxh.shape)
print("Whh:", Whh.shape)
print("Why:", Why.shape)
print("bh :", bh.shape)
print("by :", by.shape)

def propagacion_adelante_rnn(
    secuencia,
    Wxh,
    Whh,
    Why,
    bh,
    by
):

    # Estado oculto inicial
    h = np.zeros(
        (1, Whh.shape[0])
    )

    # Lista para almacenar los estados
    estados = []

    # Procesar cada elemento de la secuencia
    for t in range(len(secuencia)):

        x_t = secuencia[t].reshape(1, 1)

        # Ecuación de la RNN
        h = tanh(
            np.dot(x_t, Wxh)
            +
            np.dot(h, Whh)
            +
            bh
        )

        estados.append(h)

    # Obtener la salida final
    y_predicha = np.dot(h, Why) + by

    return y_predicha, estados

def calcular_error(y_real, y_predicha):

    return np.mean(
        (y_real - y_predicha) ** 2
    )

def entrenar_rnn(
    X,
    y,
    Wxh,
    Whh,
    Why,
    bh,
    by,
    tasa_aprendizaje,
    epocas
):

    errores = []

    # Ciclo de épocas
    for epoca in range(epocas):

        error_total = 0

        # Recorrer todas las secuencias
        for i in range(len(X)):

            secuencia = X[i]

            valor_real = y[i].reshape(1, 1)


            y_predicha, estados = propagacion_adelante_rnn(
                secuencia,
                Wxh,
                Whh,
                Why,
                bh,
                by
            )

            error = valor_real - y_predicha

            error_total = (
                error_total
                +
                np.mean(error ** 2)
            )


            # Último estado oculto
            h_final = estados[-1]

            Why = (
                Why
                +
                tasa_aprendizaje
                *
                np.dot(
                    h_final.T,
                    error
                )
            )

            by = (
                by
                +
                tasa_aprendizaje
                *
                error
            )

            ajuste_oculto = (
                np.dot(
                    error,
                    Why.T
                )
                *
                derivada_tanh(h_final)
            )


            # Último valor de la secuencia
            ultimo_valor = (
                secuencia[-1]
                .reshape(1, 1)
            )

            Wxh = (
                Wxh
                +
                tasa_aprendizaje
                *
                np.dot(
                    ultimo_valor.T,
                    ajuste_oculto
                )
            )

            bh = (
                bh
                +
                tasa_aprendizaje
                *
                ajuste_oculto
            )


        # Error promedio de la época
        error_promedio = (
            error_total
            /
            len(X)
        )

        errores.append(error_promedio)


        # Mostrar error cada 100 épocas
        if epoca % 100 == 0:

            print(
                "Época:",
                epoca,
                "| Error:",
                error_promedio
            )


    return (
        Wxh,
        Whh,
        Why,
        bh,
        by,
        errores
    )

tasa_aprendizaje = 0.01
epocas = 1000


print("\n====================================")
print("INICIANDO ENTRENAMIENTO")
print("====================================")


Wxh, Whh, Why, bh, by, errores = entrenar_rnn(

    X,
    y,

    Wxh,
    Whh,
    Why,
    bh,
    by,

    tasa_aprendizaje,
    epocas

)


print("\n====================================")
print("ENTRENAMIENTO TERMINADO")
print("====================================")

print(
    "Error inicial:",
    errores[0]
)

print(
    "Error final:",
    errores[-1]
)

pronosticos = []


for i in range(len(X)):

    y_predicha, estados = propagacion_adelante_rnn(

        X[i],

        Wxh,
        Whh,
        Why,
        bh,
        by

    )

    pronosticos.append(
        y_predicha[0, 0]
    )


pronosticos = np.array(
    pronosticos
)

plt.figure(figsize=(10, 5))

plt.plot(
    y,
    label="Serie real"
)

plt.plot(
    pronosticos,
    label="Pronóstico de la RNN"
)

plt.title(
    "Pronóstico de serie temporal con RNN"
)

plt.xlabel("Tiempo")
plt.ylabel("Valor")

plt.legend()

plt.grid(True)

plt.show()

plt.figure(figsize=(10, 5))

plt.plot(
    errores
)

plt.title(
    "Error durante el entrenamiento"
)

plt.xlabel("Épocas")

plt.ylabel(
    "Error cuadrático medio"
)

plt.grid(True)

plt.show()

# Tomar los últimos 5 valores
ultima_secuencia = (
    serie[-pasos:]
    .reshape(pasos, 1)
)


# Realizar el pronóstico
siguiente_valor, estados = propagacion_adelante_rnn(

    ultima_secuencia,

    Wxh,
    Whh,
    Why,
    bh,
    by

)

print("\n====================================")
print("PRONÓSTICO FUTURO")
print("====================================")

print(
    "Últimos valores utilizados:"
)

print(
    ultima_secuencia.flatten()
)


print(
    "\nPronóstico del siguiente valor:"
)

print(
    siguiente_valor[0, 0]
)

print(
    "\nÚltimo valor real de la serie:"
)

print(
    serie[-1]
)

print("\n====================================")
print("RESUMEN DE LA PRÁCTICA")
print("====================================")

print(
    "Número de datos:",
    len(serie)
)

print(
    "Número de secuencias:",
    len(X)
)

print(
    "Pasos utilizados:",
    pasos
)

print(
    "Neuronas ocultas:",
    neuronas_ocultas
)

print(
    "Tasa de aprendizaje:",
    tasa_aprendizaje
)

print(
    "Épocas:",
    epocas
)

print(
    "Error final:",
    errores[-1]
)

print(
    "Pronóstico del siguiente valor:",
    siguiente_valor[0, 0]
)
```
