```python
import tensorflow as tf
import numpy as np

X = np.array([[1.0,2.0,3.0]])
y = np.array([[1.0]])

modelo = Sequential([Dense(2, activation='sigmoid',input_shape=(3,)),
                     Dense(1,activation="sigmoid")])

W = np.array([[0.2,0.3],[0.4,0.5],[0.1,0.2]])
b = np.array([0.1,0.2])
V = np.array([[0.8],[0.7]])
b3 = np.array([0.1])

#Asignar pesos a las capas
modelo.layers[0].set_weights([W,b])
modelo.layers[1].set_weights([V,b3])

#Compilamos el modelo
modelo.compile(optimizer = SGD(learning_rate=0.1), loss = "mse")

print("prediccion antes")
print(modelo.predict(X))
modelo.fit(X,y,epochs=1, verbose=1)
print("prediccion despues")
print(modelo.predict(X))

for i, capa in enumerate(modelo.layers):
  print("\nCapa",i+1)
  print(capa.get_weights)
```
