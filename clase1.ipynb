# Realizar una neurona artificial de 4 entradas y la clasificacion 3 "gato", "perro", "ave" donde la probabilidad de gato sea <0.15, la probabilidad de perro sea >0.68, la probabilidad de 0.15 a 0.68 sea de "ave", x1, x2, x3. y x4 valdran 0.5,0.67,0.85, 0.32 respectivamente y los pesos w1,w2,w3,w4 seran 0.14,0.22,0.07,0.10 respectivamente y el bias(sesgo) valdra 0.09

```python
# ==========================================
# NEURONA ARTIFICIAL - 4 ENTRADAS
# ==========================================

# Entradas
x1, x2, x3, x4 = 0.5, 0.67, 0.85, 0.32

# Pesos
w1, w2, w3, w4 = 0.14, 0.22, 0.07, 0.10

# Bias (sesgo)
bias = 0.09

# Suma ponderada
suma = (x1*w1) + (x2*w2) + (x3*w3) + (x4*w4) + bias

print("Suma ponderada:", suma)

# Clasificación
if suma < 0.15:
    clase = "gato"
elif suma <= 0.68:
    clase = "ave"
else:
    clase = "perro"

print("Clasificación:", clase)
```

```python
# Entradas
x1 = 0.5
x2 = 0.67
x3 = 0.85
x4 = 0.32

# Pesos
w1 = 0.14
w2 = 0.22
w3 = 0.07
w4 = 0.10

# Sesgo
bias = 0.09

# Función de la neurona
y = (x1*w1) + (x2*w2) + (x3*w3) + (x4*w4) + bias

print(f"Resultado: {y:.4f}")

# Clasificación
if y < 0.15:
    print("Clasificación: gato")
elif y <= 0.68:
    print("Clasificación: ave")
else:
    print("Clasificación: perro")
```

```python
# Función de la neurona
def neurona(x1, x2, x3, x4, w1, w2, w3, w4, bias):

    # Suma ponderada
    y = (x1*w1) + (x2*w2) + (x3*w3) + (x4*w4) + bias

    # Clasificación
    if y < 0.15:
        return y, "gato"
    elif y <= 0.68:
        return y, "ave"
    else:
        return y, "perro"


# Entradas
x1 = 0.5
x2 = 0.67
x3 = 0.85
x4 = 0.32

# Pesos
w1 = 0.14
w2 = 0.22
w3 = 0.07
w4 = 0.10

# Sesgo
bias = 0.09

# Ejecutar la neurona
resultado, clasificacion = neurona(
    x1, x2, x3, x4,
    w1, w2, w3, w4,
    bias
)

print(f"Resultado de la neurona: {resultado:.4f}")
print(f"Clasificación: {clasificacion}")
```

```python
# Función de la neurona
def neurona(entradas, pesos, bias):

    # Inicializar la suma
    y = bias

    # Suma ponderada usando un ciclo for
    for i in range(len(entradas)):
        y += entradas[i] * pesos[i]

    # Clasificación
    if y < 0.15:
        return y, "gato"
    elif y <= 0.68:
        return y, "ave"
    else:
        return y, "perro"


# Entradas
entradas = [0.5, 0.67, 0.85, 0.32]

# Pesos
pesos = [0.14, 0.22, 0.07, 0.10]

# Sesgo
bias = 0.09

# Ejecutar la neurona
resultado, clasificacion = neurona(entradas, pesos, bias)

print(f"Resultado de la neurona: {resultado:.4f}")
print(f"Clasificación: {clasificacion}")
```

# Ejemplo: Neurona artificial (en clase)

```python
import math 
# Entradas, pesos y sesgo
x = [1.5, 0.8, -0.4]
w = [0.6, -0.3, 0.9]
b = 0.1

# Suma ponderada
z = sum(xi * wi for xi, wi in zip(x, w)) + b
print(f"Suma ponderada z = {z:.4f}")
#Funcion sigmoide
def sigmoide(z):
    return 1 / (1 + math.exp(-z))
activacion = sigmoide(z)
print(f"Activacion σ(z) = {activacion:.4f}")

#Clasificacion por umbral
clase = 1 if activacion >= 0.5 else 0
print(f"Clase predicha: {clase}")
```
